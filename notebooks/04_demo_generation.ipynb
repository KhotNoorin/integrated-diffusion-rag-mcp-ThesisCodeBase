```python
# Notebook 04 — Demo Generation

"""
Goal:
- Generate qualitative samples using the integrated model
- Save outputs for thesis figures and dashboard
"""

from pipelines.multimodal_generation import MultimodalGenerationPipeline
from utils.visualization import save_image_grid
from PIL import Image

pipeline = MultimodalGenerationPipeline()

prompts = [
    "A cinematic portrait of an astronaut exploring an alien ocean.",
    "A realistic illustration of renewable energy in futuristic cities.",
]

outputs = []
for p in prompts:
    result = pipeline.run(prompt=p)
    outputs.extend(result["images"])

save_image_grid(outputs, captions=prompts, out_path="evaluation/qualitative/thesis_figures.png")
Image.open("evaluation/qualitative/thesis_figures.png").show()
```
