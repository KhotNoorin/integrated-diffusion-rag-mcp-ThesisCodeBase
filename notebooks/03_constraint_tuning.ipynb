```python
# Notebook 03 — Constraint Tuning

"""
Goal:
- Systematically vary MCP weights
- Observe their effect on generation metrics (CLIPScore, FID)
- Log results to plots
"""

import pandas as pd
import matplotlib.pyplot as plt
from pipelines.evaluation_pipeline import EvaluationPipeline

weights = [0.5, 1.0, 1.5, 2.0]
results = []

for w in weights:
    eval_pipe = EvaluationPipeline(prompts=["A robot painting on canvas"])
    summary = eval_pipe.run_experiment(
        constraints_baseline={"factual": False},
        constraints_improved={"factual": True, "style": "realistic", "factual_weight": w},
        num_images=1,
    )
    summary["weight"] = w
    results.append(summary)

df = pd.DataFrame(results)
plt.plot(df["weight"], df["deltas_percent"].apply(lambda x: x.get("CLIPScore", 0)), marker="o")
plt.xlabel("Factual Weight")
plt.ylabel("CLIPScore Δ%")
plt.title("Impact of Factual Constraint Weight on CLIPScore")
plt.show()
```
