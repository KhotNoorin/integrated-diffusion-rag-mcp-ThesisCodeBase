```python
# Notebook 02 — Model Integration

"""
Goal:
- Load retrieval, diffusion, and constraint modules
- Test combined multimodal generation flow
- Validate end-to-end inference
"""

from models.retrieval.retriever import Retriever
from models.diffusion.base_diffusion import DiffusionModel
from models.constraints.constraint_manager import ConstraintManager
from models.multimodal_generator import MultimodalGenerator

retriever = Retriever(index_path="data/embeddings/chroma_index")
diffusion = DiffusionModel(pretrained_name="runwayml/stable-diffusion-v1-5")
constraints = ConstraintManager()

generator = MultimodalGenerator(retriever, diffusion, constraints)

prompt = "A scientist reading a holographic display in a futuristic lab."
output = generator.generate(prompt, mode="image")

output["image"].show()
print("Generated caption:", output.get("caption", "N/A"))
```
