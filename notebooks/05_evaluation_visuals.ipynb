```python
# Notebook 05 — Evaluation Visuals

"""
Goal:
- Load metrics, generate summary plots
- Combine with human evaluation results
- Export to thesis-ready visuals (LaTeX + PNG)
"""

import pandas as pd
import matplotlib.pyplot as plt

metrics = pd.read_csv("experiments/results/metrics_logs.csv")
human = pd.read_csv("evaluation/human_study/human_eval_summary.csv")

fig, ax = plt.subplots(1, 2, figsize=(10, 4))
metrics.groupby("metric")["delta_percent"].mean().plot(kind="bar", ax=ax[0], title="Metric Improvements (%)")
human.set_index("experiment")[["rating_overall"]].plot(kind="bar", ax=ax[1], title="Human Overall Ratings (1–5)")
plt.tight_layout()
plt.savefig("evaluation/reports/final_summary_plots.png")
plt.show()
```
