```python
# Notebook 01 — Data Exploration

"""
Goal:
- Load multimodal dataset (text + image)
- Explore caption distribution, class diversity
- Visualize embeddings (PCA/TSNE)
- Verify retrieval dataset quality
"""

import os
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
import numpy as np
from PIL import Image
from utils.logging_utils import get_logger
from training.dataset_loader import get_dataloader

logger = get_logger("data_exploration")

# Load dataset
dataset_path = "data/processed/multimodal_dataset"
loader = get_dataloader(dataset_path, batch_size=8)

# Preview captions
batch = next(iter(loader))
print("Sample captions:", batch["caption"][:5])

# Example image visualization
fig, ax = plt.subplots(1, 3, figsize=(8, 3))
for i in range(3):
    img = batch["image"][i].permute(1, 2, 0).numpy()
    ax[i].imshow((img * 0.5 + 0.5))
    ax[i].set_title(batch["caption"][i])
    ax[i].axis("off")
plt.show()

# TODO: Add embedding visualization once embeddings are computed.
```
